```python
def load_dataset(*src_filenames, labels=None):
    data = []
    for filename in src_filenames:
        with open(filename) as f:
            for line in f:
                d = json.loads(line)
                if labels is None or d['gold_label'] in labels:
                    data.append(d)
    return data
```

```python
# IMPORTANT: change BASE_DIR
BASE_DIR = '/content/drive/MyDrive/Explainable AI/XAI_project/'

LOCAL_DIR_R1_TRAIN = 'data/dynasent-v1.1-round01-yelp-train.jsonl'
LOCAL_DIR_R2_TRAIN = 'data/dynasent-v1.1-round02-dynabench-train.jsonl'

LOCAL_DIR_R1_VAL = 'data/dynasent-v1.1-round01-yelp-dev.jsonl'
LOCAL_DIR_R2_VAL = 'data/dynasent-v1.1-round02-dynabench-dev.jsonl'

LOCAL_DIR_R1_TEST = 'data/dynasent-v1.1-round01-yelp-test.jsonl'
LOCAL_DIR_R2_TEST = 'data/dynasent-v1.1-round02-dynabench-test.jsonl'
```

```python
ternary_labels = ('positive', 'negative', 'neutral')

r1_train = load_dataset(os.path.join(BASE_DIR, LOCAL_DIR_R1_TRAIN),  labels=ternary_labels)
r2_train = load_dataset(os.path.join(BASE_DIR, LOCAL_DIR_R2_TRAIN), labels=ternary_labels)

r1_val = load_dataset(os.path.join(BASE_DIR, LOCAL_DIR_R1_VAL),  labels=ternary_labels)
r2_val = load_dataset(os.path.join(BASE_DIR, LOCAL_DIR_R2_VAL), labels=ternary_labels)

r1_test = load_dataset(os.path.join(BASE_DIR, LOCAL_DIR_R1_TEST),  labels=ternary_labels)
r2_test = load_dataset(os.path.join(BASE_DIR, LOCAL_DIR_R2_TEST), labels=ternary_labels)
```

```python
df_train = pd.DataFrame({'text' : [d['sentence'] for d in r1_train] + [d['sentence'] for d in r2_train],
                         'round': [1] * len(r1_train) + [2] * len(r2_train),
                         'label' : [d['gold_label'] for d in r1_train] + [d['gold_label'] for d in r2_train]})

df_val = pd.DataFrame({'text' : [d['sentence'] for d in r1_val] + [d['sentence'] for d in r2_val],
                         'round': [1] * len(r1_val) + [2] * len(r2_val),
                         'label' : [d['gold_label'] for d in r1_val] + [d['gold_label'] for d in r2_val]})

df_test = pd.DataFrame({'text' : [d['sentence'] for d in r1_test] + [d['sentence'] for d in r2_test],
                         'round': [1] * len(r1_test) + [2] * len(r2_test),
                         'label' : [d['gold_label'] for d in r1_test] + [d['gold_label'] for d in r2_test]})
```
